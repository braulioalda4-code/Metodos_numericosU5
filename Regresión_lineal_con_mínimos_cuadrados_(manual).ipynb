Ejercicio 1 - Regresión lineal con mínimos cuadrados (manual)

```python
import numpy as np  # Importa la biblioteca NumPy para cálculos matriciales

x = np.array([1, 2, 3, 4, 5])  # Define los datos de entrada
y = np.array([2, 4, 5, 4, 5])  # Define los datos observados

X = np.vstack([np.ones(len(x)), x]).T  # Construye la matriz de diseño con columna de unos y val

XT = X.T  # calcula la transpuesta de la matriz X

beta = np.linalg.inv(XT.dot(X)).dot(XT).dot(y)  # Aplica la fórmula de mínimos cuadrados

b = beta[0]  # Extrae el intercepto
m = beta[1]  # Extrae la pendiente

y_pred = m*x + b  # Calcula los valores estimados por el modelo

print("Pendiente:", m)     # Muestra la pendiente
print("Intercepto:", b)    # Muestra el intercepto
print("Predicción:", y_pred)  # Muestra los valores ajustados

# Construye la matriz de diseño con columna de unos y valores de x
```
