Ejercicio 3 - Correlacion

```python
import numpy as np # importa numpy

x = np.array([1,2,3,4,5]) # Datos de variable independiente
y = np.array([2,4,5,4,5]) # Datos de variable independiente

correlacion = np.corrcoef(x,y)[0,1] # Calcula la correlación

print("Correlación:",correlacion) #Muestra resultado
```
