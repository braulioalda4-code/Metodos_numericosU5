Ejercicio 3 - Reconstrucción de datos

```python
import numpy as np  # Importa NumPy

x = np.array([1,2,3,4,5])  # Datos de entrada
y = np.array([2,4,5,4,5])  # Datos reales

coef = np.polyfit(x,y,1)   # Ajuste Lineal

y_pred = coef[0]*x + coef[1]  # Predicciones

error = np.mean((y - y_pred)**2)  # Calcula el error cuadrático medio

print("Error MSE:",error)  # Muestra el error
```
