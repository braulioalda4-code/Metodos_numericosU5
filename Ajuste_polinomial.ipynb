Ejercicio 2 - Ajuste polinomial

```python
import numpy as np  # Importa NumPy

x = np.array([1,2,3,4,5])    # Datos de entrada
y = np.array([2,4,5,4,5])    # Datos observados

coef = np.polyfit(x,y,2)     # Ajusta un polinomio de grado 2

a = coef[0]    # Coeficiente cuadrático
b = coef[1]    # Coeficiente lineal
c = coef[2]    # Término independiente

y_pred = a*x**2 + b*x + c    # Calcula valores predichos

print("Coeficientes:",coef)  # Muestra coeficientes
print("Predicción:",y_pred)  # Muestra ajuste
```
